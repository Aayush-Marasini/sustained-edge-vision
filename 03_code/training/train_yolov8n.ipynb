```python
# 1. Prepare Environment
!pip install ultralytics -q

import yaml
import torch
import random
import numpy as np
import os
from ultralytics import YOLO

# --- REPRODUCIBILITY CONFIG (IEEE/Nature standard) ---
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
print(f"🔐 Reproducibility Locked. Global Seed: {SEED}")

# 2. CONFIGURE PATHS
# Using the confirmed path with dashes
DATASET_PATH = '/kaggle/input/datasets/aayushmarasini1/usa-split-yolo'

data_config = {
    'path': DATASET_PATH,
    'train': 'train/images', 
    'val': 'val/images',     
    'nc': 4,
    'names': ['D00', 'D10', 'D20', 'D40']
}

# Overwrite the data.yaml with the corrected path
with open('data.yaml', 'w') as f:
    yaml.dump(data_config, f)

# 3. VERIFICATION
if os.path.exists(os.path.join(DATASET_PATH, 'train/images')):
    print(f"✅ Verified! Data found at: {DATASET_PATH}")
else:
    print(f"❌ Critical Error: Data still not found at {DATASET_PATH}")

# 4. Sustained Training for 300 Epochs
model = YOLO('yolov8n.pt') 

model.train(
    data='data.yaml',
    epochs=300,        # Maximum cap for model convergence [cite: 63]
    patience=50,       # Early stopping check
    imgsz=640,
    batch=16,
    device=0,
    seed=SEED,
    deterministic=True,
    project='sustained_edge_vision',
    name=f'yolov8n_baseline_seed{SEED}'
)
```
